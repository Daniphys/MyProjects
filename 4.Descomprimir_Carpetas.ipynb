```python
import os
import zipfile

def descomprimir_en_subcarpetas(ruta_principal, ruta_destino):
    """
    Descomprime todos los archivos .zip encontrados en la ruta_principal
    y guarda su contenido en una subcarpeta con el nombre del archivo.
    
    :param ruta_principal: Carpeta donde están los archivos comprimidos.
    :param ruta_destino: Carpeta donde se guardará todo el contenido descomprimido.
    """
    # Crear carpeta destino si no existe
    if not os.path.exists(ruta_destino):
        os.makedirs(ruta_destino)

    # Iterar sobre todos los archivos en la carpeta principal
    for archivo in os.listdir(ruta_principal):
        if archivo.endswith(".zip"):
            ruta_zip = os.path.join(ruta_principal, archivo)
            nombre_carpeta = archivo.replace(".zip", "")
            carpeta_destino = os.path.join(ruta_destino, nombre_carpeta)

            # Crear subcarpeta para cada archivo
            if not os.path.exists(carpeta_destino):
                os.makedirs(carpeta_destino)

            print(f"Descomprimiendo {archivo} en {carpeta_destino}...")
            
            try:
                with zipfile.ZipFile(ruta_zip, 'r') as zip_ref:
                    zip_ref.extractall(carpeta_destino)
                print(f"✔ Archivo {archivo} descomprimido correctamente.")
            except Exception as e:
                print(f"❌ Error al descomprimir {archivo}: {e}")

# Ejemplo de uso
ruta_principal = r"C:\Usuario\Numero_usuario\Nube\Carpeta donde esta guardada\...\ANULAR Y REFACTURAR\SEPTIEMBRE\ZIP Comprimidos"  # Carpeta donde están los .zip
ruta_destino   = r"C:\Usuario\Numero_usuario\Nube\Carpeta donde esta guardada\...\ANULAR Y REFACTURAR\SEPTIEMBRE\ZIP Descomprimidos" # Carpeta donde se guardará todo

descomprimir_en_subcarpetas(ruta_principal, ruta_destino)

```
